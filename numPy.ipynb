```python
import numpy as np

# Roll the dice once
dice_roll = np.random.randint(1, 7)   # 7 is exclusive
print("You rolled:", dice_roll)

# Roll the dice 10 times
rolls = np.random.randint(1, 7, size=10)
print("10 rolls:", rolls)

# Simulate 4 players rolling once each
players = np.random.randint(1, 7, size=4)
print("Players rolled:", players)
```
